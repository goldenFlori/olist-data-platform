```python
# watermark: mban daten e fundit te perpunuar per çdo run inkremental
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog_name}.{control_schema}.watermark (
    pipeline_name STRING,
    last_processed_date DATE,
    updated_at TIMESTAMP
)
""")

# run_log: statistikat e çdo ekzekutimi (kerkesa: rows read/written/quarantined/duration)
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {catalog_name}.{control_schema}.run_log (
    run_id STRING,
    layer STRING,
    table_name STRING,
    rows_read LONG,
    rows_written LONG,
    rows_quarantined LONG,
    duration_seconds DOUBLE,
    run_timestamp TIMESTAMP
)
""")

print("Control tables created: watermark, run_log")
```
