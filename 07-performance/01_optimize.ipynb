```python
# OPTIMIZE bashkon skedaret e vegjel (file compaction)
# Z-ORDER rirendit sipas kolonave qe filtrohen shpesh (partition pruning me i mire)

# fact_order_items — tabela me e madhe e fakteve, Z-ORDER sipas date dhe product
spark.sql(f"OPTIMIZE {catalog_name}.{gold_schema}.fact_order_items ZORDER BY (date_key, product_id)")

# fact_orders
spark.sql(f"OPTIMIZE {catalog_name}.{gold_schema}.fact_orders ZORDER BY (date_key)")

# fact_payments
spark.sql(f"OPTIMIZE {catalog_name}.{gold_schema}.fact_payments ZORDER BY (date_key)")

# geolocation ne silver (1M rreshta)
spark.sql(f"OPTIMIZE {catalog_name}.{silver_schema}.geolocation")

print("OPTIMIZE + Z-ORDER complete")
```

```python
import time

# nje query tipike: revenue sipas date
start = time.time()
result = spark.sql(f"""
    SELECT date_key, SUM(price) as revenue
    FROM {catalog_name}.{gold_schema}.fact_order_items
    GROUP BY date_key
""").collect()
duration = time.time() - start
print(f"Query duration after OPTIMIZE: {duration:.2f}s, rows: {len(result)}")
```
