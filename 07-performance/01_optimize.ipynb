```python
# OPTIMIZE bashkon skedaret e vegjel (file compaction)
# Z-ORDER rirendit sipas kolonave qe filtrohen shpesh (partition pruning me i mire)

# fact_order_items — tabela me e madhe e fakteve, Z-ORDER sipas date dhe product
spark.sql(f"OPTIMIZE {catalog_name}.{gold_schema}.fact_order_items ZORDER BY (date_key, product_id)")

# fact_orders
spark.sql(f"OPTIMIZE {catalog_name}.{gold_schema}.fact_orders ZORDER BY (date_key)")

# fact_payments
spark.sql(f"OPTIMIZE {catalog_name}.{gold_schema}.fact_payments ZORDER BY (date_key)")

# geolocation ne silver (1M rreshta)
spark.sql(f"OPTIMIZE {catalog_name}.{silver_schema}.geolocation")

print("OPTIMIZE + Z-ORDER complete")
```

```python
import time

# nje query tipike: revenue sipas date
start = time.time()
result = spark.sql(f"""
    SELECT date_key, SUM(price) as revenue
    FROM {catalog_name}.{gold_schema}.fact_order_items
    GROUP BY date_key
""").collect()
duration = time.time() - start
print(f"Query duration after OPTIMIZE: {duration:.2f}s, rows: {len(result)}")
```

```python
import time

# Per te matur before/after, krijojme nje kopje te pa-optimizuar te fact_order_items,
# masim query-n para dhe pas OPTIMIZE, pastaj e fshijme.
# (Faktet reale jane optimizuar tashme, keshtu perdorim nje kopje test.)
spark.sql(f"""
    CREATE OR REPLACE TABLE {catalog_name}.{gold_schema}.fact_items_test
    AS SELECT * FROM {catalog_name}.{gold_schema}.fact_order_items
""")

# BEFORE — query para OPTIMIZE
start = time.time()
spark.sql(f"""
    SELECT date_key, SUM(price) as revenue
    FROM {catalog_name}.{gold_schema}.fact_items_test
    GROUP BY date_key
""").collect()
before = time.time() - start
print(f"BEFORE optimize: {before:.2f}s")

# apliko OPTIMIZE + Z-ORDER
spark.sql(f"OPTIMIZE {catalog_name}.{gold_schema}.fact_items_test ZORDER BY (date_key, product_id)")

# AFTER — po ai query pas OPTIMIZE
start = time.time()
spark.sql(f"""
    SELECT date_key, SUM(price) as revenue
    FROM {catalog_name}.{gold_schema}.fact_items_test
    GROUP BY date_key
""").collect()
after = time.time() - start
print(f"AFTER optimize: {after:.2f}s")

# improvement
improvement = (before - after) / before * 100
print(f"Improvement: {improvement:.1f}%")

# fshi tabelen e testit
spark.sql(f"DROP TABLE {catalog_name}.{gold_schema}.fact_items_test")
```
