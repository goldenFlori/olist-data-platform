```python
from pyspark.sql import functions as F

products = spark.table(f"{catalog_name}.{silver_schema}.products")

# merr produktet nga silver, heq kolonat qe sna duhen ne dimension
dim_product = products.select(
    "product_id",
    F.col("product_category_name_english").alias("product_category"),
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
)

(dim_product.write.format("delta").mode("overwrite").option("overwriteSchema", "true")
    .saveAsTable(f"{catalog_name}.{gold_schema}.dim_product"))
print(f"Wrote {dim_product.count():,} products")
```
